```python
# Pearson Test (Combined_Academic_Index vs Target)
# Importing libraries
import pandas as pd
import numpy as np
from scipy.stats import pearsonr

# Load the dataset
dataset = pd.read_csv("StudentDropout.csv")  # Adjust path as needed

# Bootstrapping settings
sample_size = 50  # Resampling size
n_bootstraps = 1000  # Number of bootstrap samples

# Convert 'Target' to numeric to ensure compatibility with pearsonr
dataset['Target'] = dataset['Target'].map({'Dropout': 0, 'Graduate': 1, 'Enrolled': 2})

# Drop missing values from relevant columns
data_pearson = dataset[['Combined_Academic_Index', 'Target']].dropna()

# Bootstrapping procedure
pearson_r_values = []
pearson_p_values = []

# Bootstrapping loop
for _ in range(n_bootstraps):
    sample = data_pearson.sample(n=sample_size, replace=True)  # Resampling with replacement
    r, p = pearsonr(sample['Combined_Academic_Index'], sample['Target'])
    pearson_r_values.append(r)
    pearson_p_values.append(p)

# Sample size is fixed, so degree of freedom can be calculated outside the loop
df_pearson = sample_size - 2

# Compute mean and confidence interval
mean_r_pearson = np.mean(pearson_r_values)
mean_p_pearson = np.mean(pearson_p_values)

# Print results
print("\nPearson Correlation Test (Bootstrapped Results):")
print(f"Mean Pearson Correlation Coefficient (r): {mean_r_pearson:.4f}")
print(f"Mean P-value: {mean_p_pearson:.4f}")
print (f"Degrees of freedom: {df_pearson}")

# Interpretation
if mean_p_pearson < 0.05:
 print("\nConclusion: Since the p-value is less than 0.05, we reject the null hypothesis.")
 print("There is a statistically significant correlation between Combined_Academic_Index and Target.")
else:
  print("\nConclusion: Since the p-value is greater than 0.05, we accept the null hypothesis.")
  print("There is no statistically significant correlation between Combined_Academic_Index and Target.")
```

```python
# Spearman Test (Combined_Academic_Index  vs Target)
# Importing libraries
import pandas as pd
import numpy as np
from scipy.stats import spearmanr
import matplotlib.pyplot as plt

# Load dataset
dataset = pd.read_csv("StudentDropout.csv")

# Map 'Target' to ordinal values
dataset['Target'] = dataset['Target'].map({'Dropout': 0, 'Graduate': 1, 'Enrolled': 2})

# Drop missing values from relevant columns
data_spearman = dataset[['Combined_Academic_Index', 'Target']].dropna()

# Bootstrapping procedure
sample_size = 50
n_bootstraps = 1000

# (Bootstrapping procedure)- Lists to store bootstrapped correlation values
spearman_r_values = []
spearman_p_values = []

# Bootstrapping loop
for _ in range(n_bootstraps):
    sample = data_spearman.sample(n=sample_size, replace=True)
    r, p = spearmanr(sample['Combined_Academic_Index'], sample['Target'])
    spearman_r_values.append(r)
    spearman_p_values.append(p)

# Sample size is fixed, so degree of freedom can be calculated outside the loop
df_spearman = sample_size - 2

# Compute stats
mean_r_spearman = np.mean(spearman_r_values)
mean_p_spearman = np.mean(spearman_p_values)

# Print the results
print("\nBootstrapped Spearman Correlation Test:")
print(f"Mean Spearman Correlation Coefficient (r): {mean_r_spearman:.4f}")
print(f"Mean P-value: {mean_p_spearman:.4f}")
print (f"Degrees of freedom: {df_spearman}")

# Interpretation
if mean_p_spearman < 0.05:
 print("\nConclusion: Since the p-value is less than 0.05, we reject the null hypothesis.")
 print("There is a statistically significant correlation between Combined_Academic_Index and Target.")
else:
  print("\nConclusion: Since the p-value is greater than 0.05, we accept the null hypothesis.")
  print("There is no statistically significant correlation between Combined_Academic_Index and Target.")
```

```python
# Importing libraries
import pandas as pd
from scipy.stats import chi2_contingency

# Load the dataset
dataset = pd.read_csv("StudentDropout.csv")

# Take a sample of the dataset
sample_size = 50
sample = dataset.sample(n=sample_size, replace=False, random_state=42)

# Create a contingency table (cross-tabulation of Target vs Scholarship_Holder)
contingency_data = pd.crosstab(sample['Target'], sample['Scholarship holder'])

# Perform Chi-Square test
r, p, dof, expected = chi2_contingency(contingency_data)

# Print results
print("Chi-Square Test Results:")
print(f"Chi-Square Statistic: {r:.3f}")
print(f"P-value: {p:.3f}")
print(f"Degrees of Freedom: {dof}")

# Interpretation
if p > 0.05:
    print("Conclusion: The categories are independent (no significant association).")
else:
    print("Conclusion: The categories are dependent (significant association).")
```

```python
# T-test

import pandas as pd
from scipy.stats import ttest_1samp, ttest_ind, ttest_rel
# Load the updated dataset
dataset = pd.read_csv('StudentDropout.csv')

# One-Sample T-Test
one_sample_data = dataset['Previous qualification (grade)'].dropna()
t_stat, p_val = ttest_1samp(one_sample_data, 120)
print("One-Sample T-Test:")
print(f"T-Statistic: {t_stat:.3f}")
print(f"P-Value: {p_val:.10e}")

# Independent Two-Sample T-Test
graduate = dataset[dataset['Target'] == 'Graduate']['Curricular units 2nd sem (grade)'].dropna()
dropout = dataset[dataset['Target'] == 'Dropout']['Curricular units 2nd sem (grade)'].dropna()
t_stat, p_val = ttest_ind(graduate, dropout, equal_var=False)
print("\nIndependent Two-Sample T-Test:")
print(f"T-Statistic: {t_stat:.3f}")
print(f"P-Value: {p_val:.10e}")

# Paired T-Test
enrolled = dataset['Curricular units 2nd sem (enrolled)'].dropna()
approved = dataset['Curricular units 2nd sem (approved)'].dropna()

# Ensure same length for pairing
min_len = min(len(enrolled), len(approved))
enrolled = enrolled[:min_len]
approved = approved[:min_len]
paired_t_stat, paired_p_val = ttest_rel(enrolled, approved)
print("\nPaired T-Test:")
print(f"T-Statistic: {paired_t_stat:.3f}")
print(f"P-Value: {paired_p_val:.10e}")
```

```python
# Student Clustering using K-Means and PCA

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Step 1: Load the dataset
dataset = pd.read_csv('StudentDropout.csv')

# Step 2: Preprocess the data
features = dataset [[
    'Admission grade',
    'Curricular units 1st sem (grade)',
    'Curricular units 2nd sem (grade)',
    'Previous qualification (grade)']].copy()
features.fillna(features.mean(), inplace=True)

# Step 3: Scale the features
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)

# Step 4: Determine optimal k using the Elbow Method
wcss = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(scaled_features)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), wcss, marker='o')
plt.title("Elbow Method for Optimal K")
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia (WCSS)")
plt.grid(True)
plt.show()

# Step 5: Apply KMeans
k = 3  # Chosen from elbow method (you can adjust based on the Elbow graph)
print(f"Chosen number of clusters (k): {k}")
kmeans = KMeans(n_clusters=k, random_state=42)
cluster_labels = kmeans.fit_predict(scaled_features)
dataset['Cluster'] = cluster_labels

# Step 6: PCA for 2D Visualization
pca = PCA(n_components=2)
pca_components = pca.fit_transform(scaled_features)
dataset['PCA1'] = pca_components[:, 0]
dataset['PCA2'] = pca_components[:, 1]

plt.figure(figsize=(8, 6))
sns.scatterplot(x='PCA1', y='PCA2', hue='Cluster', data=dataset, palette='Set2', s=100)
plt.title("Clusters of Students (PCA Reduced to 2D)")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend(title='Cluster')
plt.grid(True)
plt.show()
```

```python
# Student Clustering using Hierarchical Clustering (Agglomerative Clustering)

import pandas as pd
import matplotlib.pyplot as plt
import scipy.cluster.hierarchy as sch
from sklearn.cluster import AgglomerativeClustering

# Load the dataset
dataset = pd.read_csv('StudentDropout.csv')

# Selecting the relevant features for clustering (Academic Performance)
X = dataset[[
    "Admission grade",
    "Curricular units 1st sem (grade)",
    "Curricular units 2nd sem (grade)",
    "Previous qualification (grade)"
]].values

# Step 1: Create Dendrogram to determine optimal number of clusters
plt.figure(figsize=(10, 7))
dendrogram = sch.dendrogram(sch.linkage(X, method='ward'))
plt.title("Dendrogram")
plt.xlabel("Students")
plt.ylabel("Euclidean distances")
plt.grid(True)
plt.tight_layout()
plt.show()

# Step 2: Apply Agglomerative Clustering (choose 3 clusters based on dendrogram)
hc = AgglomerativeClustering(n_clusters=3, metric='euclidean', linkage='ward')
y_hc = hc.fit_predict(X)

# Step 3: Visualize the clusters (using first two features for 2D plot)
plt.figure(figsize=(10, 7))
colors = ['red', 'blue', 'green']
labels = ['Cluster 1', 'Cluster 2', 'Cluster 3']

for i in range(3):
    plt.scatter(X[y_hc == i, 0], X[y_hc == i, 1], s=100, c=colors[i], label=labels[i])

plt.title("Clusters of Students (Based on Academic Grades)")
plt.xlabel("Admission Grade")
plt.ylabel("Curricular units 1st sem (Grade)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

print("Cluster visualization completed.")
print("Interpretation:")
print("- Cluster 1 (Red): Students with high admission grades but varying semester performance.")
print("- Cluster 2 (Blue): Students with moderate academic scores overall.")
print("- Cluster 3 (Green): Students needing academic support (lower grades).")
```

```python
# Mean Shift Clustering

import pandas as pd
import numpy as np
from sklearn.cluster import MeanShift
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# Load the dataset
data = pd.read_csv('StudentDropout.csv')

# Step 1: Data Preprocessing
features = [ 'Admission grade', 'Curricular units 1st sem (grade)',
            'Curricular units 2nd sem (grade)']

# Extract selected features
X = data[features]

# Handle missing values if any (this dataset appears complete)
X = X.dropna()

# Step 2: Feature Scaling
# Mean Shift is distance-based, so scaling is important
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 3: Mean Shift Clustering
# Initialize Mean Shift - bandwidth affects the number of clusters
# If not specified, sklearn will estimate it automatically
ms = MeanShift(bandwidth=None, bin_seeding=True)

# Fit the model to our data
ms.fit(X_scaled)

# Get the cluster labels for each data point
labels = ms.labels_

# Step 4: Analyze Results
# Number of clusters found
n_clusters = len(np.unique(labels))
print(f"Number of estimated clusters: {n_clusters}")

# Add cluster labels back to original data for analysis
data['Cluster'] = labels

# Step 5: Visualize Clusters (using PCA for dimensionality reduction)
# Reduce to 2 dimensions for visualization
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Create scatter plot
plt.figure(figsize=(12, 8))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=labels, cmap='viridis', alpha=0.6)

# Add legend and labels
plt.legend(*scatter.legend_elements(), title="Clusters")
plt.title('Mean Shift Clustering Results (PCA-reduced)')
plt.xlabel('PCA Component 1')
plt.ylabel('PCA Component 2')
plt.show()

# Step 6: Analyze Cluster Characteristics
# Calculate mean values for each feature by cluster
cluster_analysis = data.groupby('Cluster')[features].mean()
print("\nCluster characteristics:")
print(cluster_analysis)

# Step 7: Examine Target Variable Distribution in Clusters
# See how dropout status distributes across clusters
if 'Target' in data.columns:
    target_distribution = pd.crosstab(data['Cluster'], data['Target'], normalize='index')
    print("\nTarget variable distribution by cluster:")
    print(target_distribution)
```
